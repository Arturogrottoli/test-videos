# Coffee shop sales analysis

Upload `sales.csv` to the Files panel (folder icon on the left) before running the cells.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")
df.head()
```

```python
df.describe()
```

```python
by_product = df.groupby("product")["revenue"].sum()
by_product
```

```python
by_product.plot(kind="bar", title="Revenue by product", rot=0)
plt.ylabel("Revenue ($)")
plt.show()
```
